```python
import pandas as pd

df = pd.read_csv(r"C:\Users\jagad\OneDrive\Documents\emissions.csv")

rename_columns = {
    'Emissions.Sector.Power Industry': 'Power Emissions',
    'Emissions.Sector.Transport': 'Transport Emissions',
    'Emissions.Sector.Buildings': 'Buildings Emissions',
    'Emissions.Sector.Other Industry': 'Other Industry Emissions',
    'Emissions.Sector.Other sectors': 'Other Sector Emissions',
    'Emissions.Type.CO2': 'Total CO2 Emissions',
    'Emissions.Type.CH4': 'Total CH4 Emissions',
    'Emissions.Type.N2O': 'Total N2O Emissions',
    'Ratio.Per GDP': 'Emissions per GDP',
    'Ratio.Per Capita': 'Emissions per Capita'
}
df.rename(columns=rename_columns, inplace=True)

df.to_csv(r"C:\Users\jagad\OneDrive\Documents\emissions_cleaned.csv", index=False)

 #Converting to long format (melt sector columns)

df_melted = df.melt(
    id_vars=['Country', 'Year'],
    value_vars=[
        'Power Emissions',
        'Transport Emissions',
        'Buildings Emissions',
        'Other Industry Emissions',
        'Other Sector Emissions'
    ],
    var_name='Sector',
    value_name='Emissions'
)

df_melted.to_csv(r"C:\Users\jagad\OneDrive\Documents\emissions_long_format.csv", index=False)
```
